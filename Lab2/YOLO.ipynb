```python
from ultralytics import YOLO

# load nano model (fastest)
model = YOLO("yolov8n.pt")

print("YOLOv8 model loaded")
```

```python
# Run detection on video and save result
results = model.predict(
    source="/content/zebra.mp4",  # your uploaded video path
    save=True,                    # save output video
    conf=0.25                     # confidence threshold
)

print("Detection complete!")
```

```python
import os
from IPython.display import HTML
from base64 import b64encode

output_dir = "/content/runs/detect/predict"

files_in_dir = os.listdir(output_dir)
print("Files found in output folder:", files_in_dir)

# look for any video file (mp4 or avi)
video_files = [f for f in files_in_dir if f.endswith((".mp4", ".avi"))]

if len(video_files) == 0:
    print("No video file found. Make sure detection cell ran successfully.")
else:
    video_path = os.path.join(output_dir, video_files[0])
    print("Displaying:", video_path)

    mp4 = open(video_path,'rb').read()
    data_url = "data:video/mp4;base64," + b64encode(mp4).decode()

    HTML(f"""
    <video width=600 controls>
          <source src="{data_url}" type="video/mp4">
    </video>
    """)
```
